# AI Image Retrieval - Experiments
This notebook demonstrates the core components of the system interactively.

```python
import sys
from pathlib import Path
sys.path.append(str(Path().resolve().parent))

from src.config import config
from src.search_engine import SearchEngine
from PIL import Image
import matplotlib.pyplot as plt

engine = SearchEngine()
```

```python
query = "a person walking a dog"
results = engine.search(query, top_k=3)

fig, axes = plt.subplots(1, len(results), figsize=(15, 5))
if len(results) == 1: axes = [axes]
for ax, res in zip(axes, results):
    img = Image.open(res['metadata']['file_path'])
    ax.imshow(img)
    ax.set_title(f"Score: {res['final_score']:.2f}")
    ax.axis('off')
plt.show()
```
